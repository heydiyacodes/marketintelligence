```python
# SECTION 1: Load Data
import pandas as pd
import numpy as np
import psycopg2
from datetime import datetime

# Load telecom churn data
churn_data = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print(f"✅ Loaded churn data: {churn_data.shape}")
print(f"\nFirst few rows:")
print(churn_data.head())

print(f"\nData types:")
print(churn_data.dtypes)

print(f"\nMissing values:")
print(churn_data.isnull().sum())

# SECTION 2: Create Binary Churn Variable
churn_data['Churn_binary'] = (churn_data['Churn'] == 'Yes').astype(int)
print(f"\n✅ Churn Distribution:")
print(churn_data['Churn_binary'].value_counts())
print(f"Churn rate: {churn_data['Churn_binary'].mean():.1%}")

# SECTION 3: Map to Beauty Categories
churn_data['primary_category'] = churn_data['InternetService'].map({
    'Fiber optic': 'Skincare',
    'DSL': 'Makeup',
    'No': 'Accessories'
})

churn_data['secondary_category'] = churn_data['PhoneService'].map({
    'Yes': 'Wellness',
    'No': 'Haircare'
})

print(f"\n✅ Primary Category Distribution:")
print(churn_data['primary_category'].value_counts())

# SECTION 4: Create Customer Segments
churn_data['segment'] = pd.cut(
    churn_data['MonthlyCharges'],
    bins=[0, 30, 60, 100, 150],
    labels=['Budget', 'Mid-tier', 'Premium', 'Enterprise']
)

print(f"\n✅ Segment Distribution:")
print(churn_data['segment'].value_counts())

# Save processed data
churn_data.to_csv('churn_data_processed.csv', index=False)
print(f"\n✅ Saved to churn_data_processed.csv")
```
